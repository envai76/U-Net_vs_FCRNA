```python
import os
import cv2
import numpy as np
import csv

centroids_folder = './synth_data/dots/'

def load_images_from_folder(input_image_folder , label_image_folder , type_of_label , factor= 100):
    images = []
    labels = []
    filename_counts = []
    for filename in os.listdir(input_image_folder):
        # print(filename , filename[:-8] + 'dots.png')
        img_path = os.path.join(input_image_folder, filename)
        img = cv2.imread(img_path)  # Read as grayscale
        label_path = os.path.join(label_image_folder, filename[:-7] + 'dots.png')
        label_image = cv2.imread( label_path )
        # cv2_imshow(label_image)
        if type_of_label== 'center':
          label_image= factor * (label_image[:,:,2] > 0)
        label_image = np.expand_dims(label_image, axis=-1)
        # print(label_image.shape)  # Output: (256, 256, 1)
        # print(np.unique(label_image))
        print(filename[:-7] + 'dots.png', str(np.sum(label_image == 1)))
        counts=np.sum(label_image == 1)
        filename_counts.append((filename[:-7] + 'dots.png', str(np.sum(label_image == 1))))

        if img is not None and label_image is not None :
            # print(img.shape , np.unique(label_image) )
            images.append(img)
            labels.append(label_image)

        # Write filename and count data to CSV file
        csv_file= "./synth_data/outputs/"+filename[:-7]+"/true_counts.txt"
        # csv_file = 'true_counts.csv'
        print(csv_file)
        with open(csv_file, 'w', newline='') as file:
            file.write(str(counts))


    print(f'CSV file "{csv_file}" has been created successfully.')

    return np.array(images), np.array(labels)

input_image_folder= './synth_data/dics/'
label_center_folder= './synth_data/dots/'

images, labels = load_images_from_folder(input_image_folder , label_center_folder , "center" , factor = 1)
```
